03] Rules for declaring variables

```python
# Here are the rules for declaring variables in Python explained in simple words:

# Start with a letter or underscore: A variable name must begin with a letter (a–z, A–Z) or an underscore (_). 
# It cannot start with a number.Example: age = 25 or _status = "Active"


# No spaces or special characters: Variable names can only contain letters, numbers, and underscores. 
# You cannot use spaces, hyphens, or punctuation.Example: Use user_name = "Alex" instead of user name = "Alex".


# Case-sensitive: Capital letters matter. Python treats uppercase and lowercase names as completely different
#  variables.Example: score = 10 and Score = 20 are two different variables.


# Cannot use Python keywords: You cannot use words that Python already reserves for its own logic
# (like if, else, print, for, while, or import).Example: You cannot write if = 5. Instead, use if_condition = 5.


# Can contain numbers (just not at the start): You can freely use numbers anywhere else inside the variable name
# Example: player1_score = 95
```

```python
# ==========================================
# 1. ARITHMETIC OPERATORS (Math)
# ==========================================
print("--- Arithmetic Operators ---")
a, b = 10, 3

print(f"Addition (+):         10 + 3 = {a + b}")
print(f"Subtraction (-):      10 - 3 = {a - b}")
print(f"Multiplication (*):   10 * 3 = {a * b}")
print(f"Division (/):         10 / 3 = {a / b}")  # Always gives a float (decimal)

print(f"Floor Division (//):  10 // 3 = {a // b}")  # Divides and rounds down
print(f"Modulus (%):          10 % 3 = {a % b}")  # Gives the remainder
print(f"Exponent (**):        10 ** 3 = {a ** b}")  # 10 to the power of 3

# ==========================================
# 2. COMPARISON OPERATORS (True or False)
# ==========================================
print("\n--- Comparison Operators ---")
x, y = 5, 8

print(f"Equal to (==):              5 == 8 is {x == y}")
print(f"Not equal to (!=):          5 != 8 is {x != y}")
print(f"Greater than (>):           5 > 8 is {x > y}")
print(f"Less than (<):              5 < 8 is {x < y}")
print(f"Greater than or equal (>=): 5 >= 8 is {x >= y}")
print(f"Less than or equal (<=):    5 <= 8 is {x <= y}")

# ==========================================
# 3. LOGICAL OPERATORS (Combining Conditions)
# ==========================================
print("\n--- Logical Operators ---")
is_sunny = True
is_warm = False

# and: True only if BOTH are true
print(f"Logical AND: is_sunny and is_warm = {is_sunny and is_warm}")
# or: True if AT LEAST ONE is true
print(f"Logical OR:  is_sunny or is_warm  = {is_sunny or is_warm}")
# not: Reverses the result
print(f"Logical NOT: not is_sunny         = {not is_sunny}")

# ==========================================
# 4. ASSIGNMENT OPERATORS (Storing Values)
# ==========================================
print("\n--- Assignment Operators ---")
num = 5
print(f"Assign (=):      num = {num}")

num += 3  # Same as num = num + 3
print(f"Add & Assign (+=): num is now {num}")

num -= 2  # Same as num = num - 2
print(f"Sub & Assign (-=): num is now {num}")

# ==========================================
# 5. MEMBERSHIP OPERATORS (Checking inside items)
# ==========================================
print("\n--- Membership Operators ---")
my_list = [1, 2, 3, 4, 5]

print(f"In (in):          3 in list     -> {3 in my_list}")
print(f"Not in (not in):  10 not in list -> {10 not in my_list}")

# ==========================================
# 6. IDENTITY OPERATORS (Checking memory location)
# ==========================================
print("\n--- Identity Operators ---")
list1 = [1, 2]
list2 = [1, 2]
list3 = list1  # Points to the exact same object

print(f"Is (is):     list1 is list3     -> {list1 is list3}")  # True (same object)
print(f"Is (is):     list1 is list2     -> {list1 is list2}")  # False (equal values, but different objects)
print(f"Is not (is not): list1 is not list2 -> {list1 is not list2}")
```

```python
e = True or False
print(f"Logical OR with True/False: True or False = {e}")#Either True --> returns True 
f = False and True
print(f"Logical AND with True/False: False and True = {f}")#Either False --> returns False 
```

```python
#not operators
print(not True)  # Returns False
print(not False)  # Returns True
```

# Typecasting in python

```python
a = 31
b = type(a)
print("Data type of a is :",b)
c = "Shams"
d = type(c)
print("Data type of c is :",d  )
```

```python
# =====================================================================
# 1. IMPLICIT TYPECASTING (Python does this automatically)
# =====================================================================
print("--- Implicit Typecasting ---")

num_int = 10  # Integer
num_float = 5.5  # Float

# Python automatically converts the integer to a float before adding
result = num_int + num_float

print(f"Result: {result}")
print(f"Data type of result: {type(result)}")  # Output will be <class 'float'>


# =====================================================================
# 2. EXPLICIT TYPECASTING (You force the conversion manually)
# =====================================================================
print("\n--- Explicit Typecasting ---")

# --- Convert to Integer using int() ---
price_str = "150"  # String representation of a number
price_int = int(price_str)  # Converts string "150" to integer 150
print(f"int('150') -> {price_int} (Type: {type(price_int)})")

decimal_num = 9.99
truncated_num = int(decimal_num)  # Chops off the decimal part (does not round!)
print(f"int(9.99)  -> {truncated_num} (Type: {type(truncated_num)})")


# --- Convert to Float using float() ---
age_int = 25
age_float = float(age_int)  # Converts 25 to 25.0
print(f"float(25)  -> {age_float} (Type: {type(age_float)})")


# --- Convert to String using str() ---
score = 98
# Useful when you want to concatenate (join) numbers with text
score_message = "Your score is " + str(score)
print(f"str(98)    -> '{score_message}' (Type: {type(score_message)})")


# --- Convert to Boolean using bool() ---
# In Python, 0, empty strings "", and empty lists [] are False. Everything else is True.
print(f"bool(1)    -> {bool(1)}")  # True
print(f"bool(0)    -> {bool(0)}")  # False
print(f"bool('Hi') -> {bool('Hi')}")  # True
print(f"bool('')   -> {bool('')}")  # False
```
