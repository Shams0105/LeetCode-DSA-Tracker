Problem 4 : Search online & write a py program to print the contents of a direcory using the os modulde

```python
#Used to print the contents of a directory using the os module
import os


def list_directory_contents(directory_path="."):
    """Prints the contents of a directory using the os module."""
    try:
        # Get all files and folders in the specified directory
        # If no path is provided, it defaults to the current working directory (".")
        contents = os.listdir(directory_path)

        print(f"\n📁 Contents of directory: '{os.path.abspath(directory_path)}'\n")

        if not contents:
            print("  (The directory is empty)")
            return

        # Loop through and display each item
        for item in contents:
            # Create the full path to check if it is a file or a folder
            full_path = os.path.join(directory_path, item)

            if os.path.isdir(full_path):
                print(f"  [FOLDER]  {item}")
            else:
                print(f"  [FILE]    {item}")

    except FileNotFoundError:
        print(f"❌ Error: The directory '{directory_path}' does not exist.")
    except PermissionError:
        print(
            f"❌ Error: You do not have permission to access '{directory_path}'."
        )


# --- Execution ---
if __name__ == "__main__":
    # Option A: List contents of the current folder where this script is running
    list_directory_contents()

    # Option B: You can also specify an absolute path like this:
    # list_directory_contents("C:\\Users\\Hello\\OneDrive\\Desktop\\DSA")
```

```python
#Optimized DSA solution abve was using loops hence not optimized , Is solved on my own
import math

primeNO = int(input("Enter your number: "))

# 1. Handle edge cases immediately
if primeNO <= 1:
    print("Not prime")
else:
    is_prime = True
    
    # 2. Loop only up to the square root of the number (inclusive)
    # math.isqrt finds the integer square root efficiently
    for i in range(2, math.isqrt(primeNO) + 1):
        if primeNO % i == 0:
            is_prime = False  # Found a factor!
            break             # Exit the loop instantly (saves thousands of operations)
            
    # 3. Check the flag variable
    if is_prime:
        print("Prime")
    else:
        print("Not prime")
```

```python
n = 3

for i in range(1, n + 1):
    # 1. Calculate the spaces needed on the left
    spaces = " " * (n - i)
    
    # 2. Calculate the stars needed for that row
    stars = "*" * (2 * i - 1)
    
    # 3. Print them joined together
    print(spaces + stars)

# Row 1: 3 - 1 = 2 spaces
# Row 2: 3 - 2 = 1 spaces
# Row 3: 3 - 3 = 0 spaces

# Stars Formula (2 * i - 1): This is the standard formula for generating consecutive odd numbers.
# Row 1: 2(1) - 1 = 1 star
# Row 2: 2(2) - 1 = 3 stars
# Row 3: 2(3) - 1 = 5 stars
```

```python
#7 Program to print the following * pattern
#   * 
#  ***
# *****  for n = 3

n = int(input())

# The outer loop manages the rows
for i in range(1, n + 1):
    
    # Nested Loop 1: Handles printing the leading spaces
    for j in range(n - i):
        print(" ", end="")
        
    # Nested Loop 2: Handles printing the stars
    for j in range(2 * i - 1):
        print("*", end="")
        
    # Move to a new line after printing spaces and stars for the current row
    print()
```

```python
#9 to print
# * * *
# *   *
# * * *
n = int(input()) # n = 3

# The outer loop manages rows (i)
for i in range(n):
    # The inner loop manages columns (j)
    for j in range(n):
        # Condition: Check if we are on any of the outer boundary edges
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("* ", end="") # Print star with a space for spacing consistency
        else:
            print("  ", end="") # Print two spaces for the hollow inner core
            
    # Move to the next line after completing the row columns
    print()

# i == 0: Represents the entire top wall.
# i == n - 1: Represents the entire bottom wall.
# j == 0: Represents the entire left wall.
# j == n - 1: Represents the entire right wall.


```
