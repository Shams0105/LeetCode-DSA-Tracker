Problem 4 : Search online & write a py program to print the contents of a direcory using the os modulde

```python
#Used to print the contents of a directory using the os module
import os


def list_directory_contents(directory_path="."):
    """Prints the contents of a directory using the os module."""
    try:
        # Get all files and folders in the specified directory
        # If no path is provided, it defaults to the current working directory (".")
        contents = os.listdir(directory_path)

        print(f"\n📁 Contents of directory: '{os.path.abspath(directory_path)}'\n")

        if not contents:
            print("  (The directory is empty)")
            return

        # Loop through and display each item
        for item in contents:
            # Create the full path to check if it is a file or a folder
            full_path = os.path.join(directory_path, item)

            if os.path.isdir(full_path):
                print(f"  [FOLDER]  {item}")
            else:
                print(f"  [FILE]    {item}")

    except FileNotFoundError:
        print(f"❌ Error: The directory '{directory_path}' does not exist.")
    except PermissionError:
        print(
            f"❌ Error: You do not have permission to access '{directory_path}'."
        )


# --- Execution ---
if __name__ == "__main__":
    # Option A: List contents of the current folder where this script is running
    list_directory_contents()

    # Option B: You can also specify an absolute path like this:
    # list_directory_contents("C:\\Users\\Hello\\OneDrive\\Desktop\\DSA")
```
