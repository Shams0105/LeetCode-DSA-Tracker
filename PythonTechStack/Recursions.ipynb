RECURSION : a function which calls itself again & again until a base codition is met

```python
#Factorial of a no using recursion
def fact(n):
    #BASE CONDITION : V imp ow the recursion will run indefinitely if we don't have a base condition
    if n == 0 or n == 1:
        return 1
    else:
        return n * fact(n-1)

n = int(input("Enter a number to calculate its factorial: "))
fact_result = fact(n)
print(f"The factorial of {n} is: {fact_result}")
```
