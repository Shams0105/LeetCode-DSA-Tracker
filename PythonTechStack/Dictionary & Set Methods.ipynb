# Mehtods in Dictionary

Dictionary

Definition:
A dictionary is a mutable, unordered collection of data stored in key-value pairs.
 
Important points
 
Data is stored as key: value pairs.
 
Keys must be unique; values can be duplicated.
 
Keys must be of an immutable/hashable type such as str, int, or tuple.
 
Dictionaries are mutable → elements can be added, changed, or removed.
 
Can access values using their Kyes

Dictionary preserves insertion order in modern Python.
 
Useful when data needs to be accessed using a meaningful key instead of an index.


```python
# Creating a dictionary
student = {
    "name": "Shams",
    "age": 20,
    "branch": "CSE"
}

# get() -> gets value using key
print(student.get("name"))          # Shams

# keys() -> returns all keys
print(student.keys())               # dict_keys(['name', 'age', 'branch'])

# values() -> returns all values
print(student.values())             # dict_values(['Shams', 20, 'CSE'])

# items() -> returns key-value pairs
print(student.items())              # dict_items([('name', 'Shams'), ...])

# update() -> adds/changes key-value pairs
student.update({"age": 21, "city": "Kolhapur"})
print(student)

# pop() -> removes specified key
student.pop("city")
print(student)

# popitem() -> removes the last inserted pair
student.popitem()
print(student)

# setdefault() -> adds key only if it doesn't exist
student.setdefault("email", "shams@gmail.com")
print(student)

# clear() -> removes everything
student.clear()
print(student)
```

SUMMARY : 

get()       → get value

keys()      → get keys

values()    → get values

items()     → get key-value pairs

update()    → add/update

pop()       → remove by key

popitem()   --> removes & returns (key , val) pair from the dict (LIFO) Last one Popped first

clear()       → clears the entire dicitionary

# Methods in Sets
SETS ARE UNORDERED

CANT CONTAIN DUPLICATES

No way to change items in sets

```python
# Creating sets
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# add() -> adds one element
A.add(7)
print(A)

# remove() -> removes element
A.remove(7)
print(A)

# discard() -> removes element, no error if absent
A.discard(10)
print(A)

# union() -> all elements from both sets
print(A.union(B))
# {1, 2, 3, 4, 5, 6}

# intersection() -> common elements
print(A.intersection(B))
# {3, 4}

# difference() -> elements in A but not in B
print(A.difference(B))
# {1, 2}

# symmetric_difference() -> elements in either set, but not both
print(A.symmetric_difference(B))
# {1, 2, 5, 6}

# update() -> adds elements from another set
A.update(B)
print(A)

# clear() -> removes all elements
A.clear()
print(A)
```

SUMMARY :-

add()                   → add element

remove()                → remove element

discard()               → remove safely

union()                 → A + B

intersection()          → common elements

difference()            → A - B

symmetric_difference()  → uncommon elements

update()                → add multiple elements

clear()                 → remove everything
