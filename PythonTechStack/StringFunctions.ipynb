# STRING FUNCTIONS

1] len(str_name) --> Outputs/Returns the kength of the string

```python
name = "MOHAMMAD AHETESHAM"
print(len(name))  # Get the length of the string
```

2]String.endwith("specify") --> this fn tells whether the variable ends with specified str/chr or not

```python
name = "MOHAMMAD AHETESHAM"
name.endswith('AM') #it is Case sensitive , so it will return False as last 2 letters aren't in caps
```

3] string.count("x") --> Counts the total no of occurences of specified(x) character

```python
name = "Mississippi"
count = name.count('i')  # Count the occurrences of the letter 'i' in the string
print(f"The letter 'i' occurs {count} times in the string '{name}'.")
```

4] string.capitalize() --> This function capitalizes the first letter of the given string

```python
name = input("Enter your name: ")
captialized_name = name.capitalize()  # Capitalize the first letter of the string
print(f"The capitalized version of your name is: '{captialized_name}'.")
```

5] string.find(word) --> this fn finds the word 7 returns the index of the first occurence of that word

```python
s = "hello world"
s.find('o')  # Find the index of the first occurrence of 'o' in the string
```
