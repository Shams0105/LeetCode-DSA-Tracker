3] string.count("x") --> Counts the total no of occurences of specified(x) character

```python
name = "Mississippi"
count = name.count('i')  # Count the occurrences of the letter 'i' in the string
print(f"The letter 'i' occurs {count} times in the string '{name}'.")
```

4] string.capitalize() --> This function capitalizes the first letter of the given string

```python
name = input("Enter your name: ")
captialized_name = name.capitalize()  # Capitalize the first letter of the string
print(f"The capitalized version of your name is: '{captialized_name}'.")
```

```python
# Define a sample string to work with
text = "  hello, Python World!  "

# 1. Modifying Case
print("--- Case Modification ---")
print(text.upper())        # "  HELLO, PYTHON WORLD!  " -> Converts all characters to uppercase
print(text.lower())        # "  hello, python world!  " -> Converts all characters to lowercase
print(text.title())        # "  Hello, Python World!  " -> Capitalizes the first letter of each word
print(text.capitalize())   # "  hello, python world!  " -> Capitalizes only the very first character

# 2. Cleaning Whitespace & Padding
print("\n--- Cleaning & Padding ---")
print(text.strip())        # "hello, Python World!"   -> Removes whitespace from both ends
print(text.lstrip())       # "hello, Python World!  " -> Removes whitespace only from the left
print(text.rstrip())       # "  hello, Python World!" -> Removes whitespace only from the right
print("42".zfill(5))       # "00042"                  -> Pads string with zeros on the left to reach width 5

# 3. Searching and Counting
# Note: We use text.strip() here to remove spaces for cleaner searching
clean_text = text.strip()  # "hello, Python World!"

print("\n--- Searching & Counting ---")
print(clean_text.count("o"))      # 3       -> Counts how many times "o" appears
print(clean_text.find("Python"))  # 7       -> Returns the starting index of the word (returns -1 if not found)
print(clean_text.index("Python")) # 7       -> Like find(), but raises a ValueError if not found
print(clean_text.startswith("he"))# True    -> Checks if string starts with "he"
print(clean_text.endswith("!"))   # True    -> Checks if string ends with "!"

# 4. Replacing and Splitting
print("\n--- Replacing & Splitting ---")
print(clean_text.replace("World", "Universe")) # "hello, Python Universe!" -> Replaces old substring with new
print(clean_text.split(", "))                  # ['hello', 'Python World!'] -> Splits string into a list by delimiter

# 5. Joining Strings
print("\n--- Joining ---")
words = ["Python", "is", "fun"]
print(" ".join(words))     # "Python is fun" -> Joins elements of a list using the string as a separator

# 6. String Checks (Booleans)
print("\n--- Content Checks ---")
print("12345".isdigit())   # True  -> Checks if all characters are numbers
print("Python".isalpha())  # True  -> Checks if all characters are alphabetic letters
print("py123".isalnum())   # True  -> Checks if all characters are alphanumeric (letters or numbers)
print("   ".isspace())     # True  -> Checks if the string contains only whitespace characters
```
