```python
# 1. Newline (\n) - Moves the text after it to a brand new line
print("--- Newline Example ---")
print("Hello\nWorld!")  
# Output:
# Hello
# World!

# 2. Horizontal Tab (\t) - Inserts a standard tab space (usually 4 spaces)
print("\n--- Tab Example ---")
print("Name:\tJohn")
print("Age:\t25")

# 3. Backslash (\\) - Prints a literal backslash character
print("\n--- Backslash Example ---")
# Used heavily for file paths in Windows
print("C:\\Users\\John\\Documents") 

# 4. Single Quote (\') and Double Quote (\") 
print("\n--- Quotes Example ---")
# Allows you to use quotes inside matching quotes without breaking the string
print('It\'s a beautiful day!')
print("The teacher said, \"Open your books.\"")

# 5. Carriage Return (\r) - Moves the cursor back to the start of the line and overwrites text
print("\n--- Carriage Return Example ---")
print("12345\rabc")  # "abc" overwrites "123"
# Output: abc45

# 6. Raw Strings (r"") - Disables all escape sequences completely
print("\n--- Raw String Bonus ---")
# Putting an 'r' before the quotes tells Python to treat backslashes as regular text
print(r"C:\Users\John\Documents\new_folder\test.txt") 
# Without 'r', \n and \t inside the path would accidentally create a newline and tab!
```
