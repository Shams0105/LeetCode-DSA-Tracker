# METHODS IN TUPLE

```python
# Create an initial tuple to demonstrate methods
# Notice it contains a few duplicate elements for testing
numbers = (10, 20, 30, 20, 40, 20, 50)
print("Initial tuple:", numbers)

# --- 1. count(x) ---
# Returns the total number of times a specific element appears in the tuple.
# If the element is not found, it returns 0.
count_of_20 = numbers.count(20)  # Returns 3 because 20 appears three times
count_of_99 = numbers.count(99)  # Returns 0 because 99 is not in the tuple

print(f"Number of times 20 appears: {count_of_20}")
print(f"Number of times 99 appears: {count_of_99}")


# --- 2. index(x) ---
# Returns the index (position) of the FIRST occurrence of a specific value.
# Note: If the item is not in the tuple, it raises a ValueError.
first_index_of_20 = numbers.index(20)  # Returns 1 (index positions start at 0)
index_of_40 = numbers.index(40)        # Returns 4

print(f"First index position of 20: {first_index_of_20}")
print(f"Index position of 40: {index_of_40}")


# --- Advanced index(x, start, end) ---
# You can optionally pass 'start' and 'end' positions to search within a specific slice.
# Let's find the second '20' by starting our search right after the first one (index 1 + 1 = 2).
second_index_of_20 = numbers.index(20, 2)  # Returns 3

print(f"The next index position of 20 (searching from index 2 onwards): {second_index_of_20}")
```

```python
# --- THE LIST USE CASE ---
# A dynamic collection of similar items that will change over time.
daily_temperatures = [32.5, 31.0, 34.2] 
daily_temperatures.append(33.1) # Perfectly valid


# --- THE TUPLE USE CASE ---
# A secure, fixed data record. 
# Changing it would corrupt the meaning of the data.
employee_record = ("Aakash", 24, "Data Engineer")

# employee_record[1] = 25 --> Throws a TypeError!


# --- THE HASHING ADVANTAGE ---
# Because tuples are immutable, they can act as complex keys in dictionaries.
locations = {}
locations[(16.70, 74.24)] = "Kolhapur office" # Valid!
# locations[[16.70, 74.24]] = "Kolhapur office" # Throws TypeError: unhashable type: 'list'
```

Some Other built-in methods to call on tuple

```python
# Create a sample tuple for demonstration
numbers = (5, 2, 8, 1, 2)

# 1. len() -> Returns the total number of items
print(len(numbers))  # Output: 5

# 2. max() and min() -> Returns the highest and lowest values
print(max(numbers))  # Output: 8
print(min(numbers))  # Output: 1

# 3. sum() -> Adds all numerical items together
print(sum(numbers))  # Output: 18

# 4. sorted() -> Returns a BRAND NEW sorted LIST (leaves the tuple untouched)
sorted_list = sorted(numbers)
print(sorted_list)  # Output: [1, 2, 2, 5, 8]
print(numbers)      # Output: (5, 2, 8, 1, 2) <-- Original tuple never changed!

# 5. tuple() -> Converts other sequences into a tuple
new_tuple = tuple([10, 20, 30])  # Converts a list into a tuple
```

# Concatenation, Repition, etc of a tuple

```python
# --- 1. BASIC CONCATENATION (+ Operator) ---
# Concatenation merges two or more tuples together.
# CRITICAL NOTE: Because tuples are immutable, this does NOT modify the original tuples.
# Instead, it creates a BRAND NEW tuple in your computer's memory.

quarters_q1 = ("Jan", "Feb", "Mar")
quarters_q2 = ("Apr", "May", "Jun")

# Merge them together using the + operator
first_half_year = quarters_q1 + quarters_q2

print("--- 1. Concatenation ---")
print(f"Q1 Tuple: {quarters_q1}")  # Stays exactly the same
print(f"Q2 Tuple: {quarters_q2}")  # Stays exactly the same
print(f"Combined New Tuple: {first_half_year}\n")


# --- 2. TUPLE REPETITION (* Operator) ---
# The * operator repeats a tuple a specific number of times.
# Just like concatenation, this creates a completely brand-new tuple.

pattern = (1, 0)
repeated_pattern = pattern * 4

print("--- 2. Repetition ---")
print(f"Original Pattern: {pattern}")
print(f"Repeated Pattern (4 times): {repeated_pattern}\n")


# --- 3. THE COMMON PITFALL: Augmented Assignment (+=) ---
# You might see developers write `my_tuple += (5, 6)`. 
# It LOOKS like it is modifying the tuple in place, but behind the scenes, 
# Python is destroying the old tuple and reassigning the variable to a new one.

coordinates = (12.5, 75.3)
print("--- 3. Augmented Assignment ---")
print(f"Original Memory ID: {id(coordinates)}")

# Appending a single element to a tuple (Note the trailing comma!)
coordinates += (99.0,) 

print(f"Updated Tuple: {coordinates}")
print(f"New Memory ID: {id(coordinates)}")  # The ID changed! It's a completely different object.

#Membership in tuple
my_tuple = (1,2,3,4,5,6)
print("\nChecks whthr specified val in tuple or not ")
print(2 in my_tuple)
print(7 in my_tuple)

print("length of my tuple is ",len(my_tuple))
```
