# Chapter 4 List Methods

```python
# Create an initial list to demonstrate methods
fruits = ["apple", "banana", "cherry"]
print("Initial list:", fruits)

# --- 1. ADDING ELEMENTS ---

# append(x): Adds an element to the absolute end of the list
fruits.append("orange")  # ['apple', 'banana', 'cherry', 'orange']

# insert(index, x): Inserts an element at a specific index position
fruits.insert(1, "mango")  # ['apple', 'mango', 'banana', 'cherry', 'orange']

# extend(iterable): Appends multiple elements from another iterable (list, tuple) to the end
tropical = ["kiwi", "papaya"]
fruits.extend(tropical)  # ['apple', 'mango', 'banana', 'cherry', 'orange', 'kiwi', 'papaya']


# --- 2. REMOVING ELEMENTS ---

# remove(x): Removes the FIRST occurrence of a specific value. Raises ValueError if not found.
fruits.remove("banana")  # ['apple', 'mango', 'cherry', 'orange', 'kiwi', 'papaya']

# pop(index): Removes and RETURNS the element at the given index. If no index, removes the last item.
popped_item = fruits.pop(2)  # Removes 'cherry' and saves it to popped_item
# List now: ['apple', 'mango', 'orange', 'kiwi', 'papaya']


# --- 3. SEARCHING & COUNTING ---

# index(x): Returns the index of the first occurrence of a value. Raises ValueError if not found.
mango_index = fruits.index("mango")  # Returns 1

# count(x): Returns the total number of times a specific element appears
fruits.append("apple")  # Add a duplicate for counting
apple_count = fruits.count("apple")  # Returns 2
fruits.pop()  # Clean up and remove that extra apple we just added


# --- 4. ORDERING & REVERSING ---

# reverse(): Reverses the elements of the list IN-PLACE (permanently updates the list)
fruits.reverse()  # ['papaya', 'kiwi', 'orange', 'mango', 'apple']

# sort(reverse=False): Sorts the items IN-PLACE (alphabetically or numerically)
fruits.sort()  # ['apple', 'kiwi', 'mango', 'orange', 'papaya'].

# Optional flag to sort descending: fruits.sort(reverse=True)


# --- 5. UTILITY & CLEANUP ---

# copy(): Returns a shallow copy of the list (safe from changing the original list variables)
fruits_backup = fruits.copy()

# clear(): Removes ALL elements from the list, leaving it completely empty
fruits.clear()  # []


# --- PRINT RESULTS TO VERIFY ---
print("\n--- Summary of Results ---")
print(f"Final cleared list (fruits): {fruits}")
print(f"Backed up list (fruits_backup): {fruits_backup}")
print(f"The item removed by pop(2) was: {popped_item}")
print(f"Index of 'mango': {mango_index}")
print(f"Count of 'apple': {apple_count}")

```
