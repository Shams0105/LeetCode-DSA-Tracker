```python
#Pattern 1
#Think of it like a grid or a table: the outer loop creates the horizontal lines (Rows), 
#and the inner loop prints the shapes across that line (Columns).
```

```python
rows = 4
columns = 4
for i in range(rows):
    for j in range (columns):  #hence the inner loop is nested inside the outer loop
        '''
        Also we connected the inner loop to the outer loop  ie when i = 0 , j will run 4(0 1 2 3)
        times and print 4 stars in the first row,
        then when i = 1, j will run 4 times again and print 4 stars in the second row and so on.
        '''
        print("*" , end = " ")

    print()  # Move to the next line after printing all columns in a row
```

```python
#Pattern 2:
rows = 5
cols = 5
for i in range(rows):
    for j in range(cols):
        if j<=i:
            print("* ", end="")
    print() #Move to the next line after printing * in all columns in a row
```
