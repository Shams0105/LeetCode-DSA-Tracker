```python
#Pattern 1
#Think of it like a grid or a table: the outer loop creates the horizontal lines (Rows), 
#and the inner loop prints the shapes across that line (Columns).
```

```python
rows = 4
columns = 4
for i in range(rows):
    for j in range (columns):  #hence the inner loop is nested inside the outer loop
        '''
        Also we connected the inner loop to the outer loop  ie when i = 0 , j will run 4(0 1 2 3)
        times and print 4 stars in the first row,
        then when i = 1, j will run 4 times again and print 4 stars in the second row and so on.
        '''
        print("*" , end = " ")

    print()  # Move to the next line after printing all columns in a row
```

```python
#Pattern 2:
rows = 5
cols = 5
for i in range(rows):
    for j in range(cols):
        if j<=i:
            print("* ", end="")
    print() #Move to the next line after printing * in all columns in a row
```

```python
#Pattern 3 :
rows = 5

cols = 5
for i in range(rows):
    nos = 1
    for j in range(cols):
        if j<=i:
            print(nos , end = " " )
            nos +=1
    print()

```

```python
#Pattern 4:
rows = 5
cols = 5
for i in range(rows):
    
    for j in range(cols):
        if j<=i:
            print(i+1, end=" ")
        
            
    print() #Move to the next line after printing all nos in all columns in a row
```

```python
'''
Pattern 5: Right-Aligned Decreasing Star Triangle
* * * * *
* * * *
* * *
* *
*

KEY IDEA: 
- Row 'i' needs exactly (cols - i) stars.
- So instead of thinking in terms of index boundaries (j <= cols-i-1),
  think in terms of "how many stars this row needs" -> stars_needed = cols - i
'''

rows = 5
cols = 5

for i in range(rows):                  # i = current row number (0-indexed)
    stars_needed = cols - i            # KEY LINE: row 0 -> 5 stars, row 1 -> 4 stars... 
    for j in range(stars_needed):      # simply loop that many times, no boundary math needed
        print("* ", end="")            # print star + space on SAME line (end="" prevents newline)
    print()                            # move to next line ONLY after finishing one full row


'''
ORIGINAL VERSION (equivalent logic, boundary-check style):
Useful to know because many problems on Striver's sheet use this 0-indexed
boundary style — j starts at 0 and goes up to (cols-1), condition decides cutoff.

for i in range(rows):
    for j in range(cols):
        if j <= (cols - i - 1):        # true only while we're within allowed star-count for this row
            print("* ", end="")
    print()
'''
```

```python
#Pattern 6:
'''
rows = 5
cols = 5
for i in range(rows):
    for j in range(cols):
        cols = rows - i
        print(j+1 , end=" ")
    print()
'''
#Pattern 6: Right-aligned decreasing number sequence
# 1 2 3 4 5
# 1 2 3 4
# 1 2 3
# 1 2
# 1

rows = 5

for i in range(rows):
    count = rows - i              # KEY FIX: calculate BEFORE the inner loop starts,
                                    # not inside it — this is the "stars_needed" idea
                                    # from before, just renamed for numbers
    for j in range(count):         # range() is now built with the CORRECT value for THIS row
        print(j + 1, end=" ")      # j+1 because j starts at 0, but we want to print from 1
    print()                        # newline after each row finishes
    
```

```python
#Pattern 7:Imp/Tougher
''''
rows = 5
cols = 9
for i in range(rows):
    
    #SPACE PRINTING
    for j in range(cols):
        if j <= rows - i - 1:
            print(" ", end=" ")

    #STAR PRINTING
    for j in range(cols):
        if j <= 2*i + 1: 
            print("*", end=" ")

    #SPACE PRINTING
    for j in range(cols):
        if j < rows - i - 1:
            print(" ", end=" ")

    print()  # Move to the next line
    '''
#Pattern 7: Pyramid (spaces + odd-number-increasing stars)
#     *
#    ***
#   *****
#  *******
# *********

rows = 5

for i in range(rows):
    
    # SPACE PRINTING
    # Row i needs (rows - i - 1) spaces on the left.
    # i=0 -> 4 spaces, i=1 -> 3 spaces ... i=4 -> 0 spaces
    for j in range(rows - i - 1):      # KEY FIX: use range() directly with the exact
        print(" ", end=" ")            # count needed — no need for an if-check at all,
       # since range() already stops at the right number
    
    # STAR PRINTING
    # Row i needs (2*i + 1) stars — the classic "odd numbers" pattern: 1,3,5,7,9
    for j in range(2 * i + 1):         # KEY FIX: same idea, range(2*i+1) auto-stops
        print("*", end=" ")            # at the correct count, no off-by-one risk
    
    print()  # move to next line after spaces + stars for this row are done
```

```python
#Pattern 8: Inverted Pyramid (spaces + odd-number-decreasing stars)
#space , stars , space
rows = 5

#
for i in range(rows):
    
    # SPACE PRINTING
    # Row i needs (rows - i - 1) spaces on the left.
    for j in range(i):                 # KEY FIX: row 0 -> 0 spaces, row 1 -> 1 space ... row 4 -> 4 spaces
        print(" ", end=" ")
    
    # STAR PRINTING
    # Row i needs (2*(rows - i) - 1) stars — the classic "odd numbers" pattern in reverse: 9,7,5,3,1
    for j in range(2 * (rows - i) - 1):   # KEY FIX: row 0 -> 9 stars, row 1 -> 7 stars ... row 4 -> 1 star
        print("*", end=" ")
    
    print()  # move to next line after spaces + stars for this row are done
```
